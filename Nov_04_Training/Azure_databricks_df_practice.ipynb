Part A : Read and Inspect the data

```python
csv_path="/Workspace/Users/azuser7403_mml.local@karthikirisoutlook.onmicrosoft.com/cinema_bookings.csv"
bookings_df=spark.read.csv(csv_path,header=True,inferSchema=True)
bookings_df.show(5)
```

```python
json_path="/Workspace/Users/azuser7403_mml.local@karthikirisoutlook.onmicrosoft.com/cinema_master.json"
cinemas_df=spark.read.json(json_path)
cinemas_df.show(5)
```

```python
bookings_df.printSchema()
cinemas_df.printSchema()
```

```python
bookings_df.show(10)
print("Number of records in bookings_df: ",bookings_df.count())
```

```python
bookings_df.select("booking_id","movie_title","tickets","booking_status").show()
```

Part B : Filter, clean and Transform

```python
confirmed_bookings = bookings_df.filter("booking_status = 'Confirmed'")
confirmed_bookings.show(5)
```

```python
filtered_bookings = bookings_df.filter("tickets > 3 AND price_per_ticket > 200")
filtered_bookings.show(5)
```

```python
from pyspark.sql.functions import col, count, when

bookings_df.select(
    count(when(col("discount_pct").isNull(), 1)).alias("missing_discount_pct"),
    count(when(col("payment_method").isNull(), 1)).alias("missing_payment_method")
).show()
```

```python
cleaned_df = bookings_df.fillna({"discount_pct": 0, "payment_method": "Unknown"})
cleaned_df.show(5)
```

```python
with_gross_df = cleaned_df.withColumn("gross_amount", col("tickets") * col("price_per_ticket"))
with_gross_df.show(5)
```

```python
with_net_df = with_gross_df.withColumn(
    "net_amount",
    col("gross_amount") * (1 - (col("discount_pct") / 100))
)
with_net_df.show(5)
```

```python
with_size_df = with_net_df.withColumn(
    "booking_size",
    when(col("tickets") >= 4, "Large").otherwise("Regular")
)
with_size_df.show(5)
```

```python
with_size_df.orderBy(col("net_amount").desc()).show(5)
```

Part C: Aggregations and joins

```python
with_size_df.groupBy("booking_status").count().show()
```

```python
from pyspark.sql.functions import col, sum as _sum

with_size_df.filter(col("booking_status") == "Confirmed") \
    .groupBy("movie_title") \
    .agg(
        _sum("tickets").alias("total_tickets"),
        _sum("net_amount").alias("net_revenue")
    ).show()
```

```python
with_size_df.filter(col("booking_status") == "Confirmed") \
    .groupBy("cinema_id") \
    .agg(_sum("net_amount").alias("confirmed_net_revenue")) \
    .orderBy(col("confirmed_net_revenue").desc()) \
    .show(3)
```

```python
joined_df = with_size_df.join(cinemas_df, on="cinema_id", how="left")
joined_df.show(5)
```

```python
joined_df.select("cinema_name", "location.city", "movie_title", "booking_status").show(5)
```

```python
joined_df.filter(col("booking_status") == "Confirmed") \
    .groupBy("location.city") \
    .agg(_sum("net_amount").alias("confirmed_net_revenue")) \
    .show()
```

```python
unmatched_df = joined_df.filter(col("cinema_name").isNull())
unmatched_df.show()
```

Part D: JSON, dates and DataFrame quality

```python
from pyspark.sql.functions import explode

exploded_facilities_df = cinemas_df.select(
    "cinema_id",
    "cinema_name",
    explode("facilities").alias("facility")
)
exploded_facilities_df.show()
```

```python
from pyspark.sql.functions import col, to_date, datediff

dates_df = joined_df.withColumn("booking_date", to_date(col("booking_date"))) \
    .withColumn("show_date", to_date(col("show_date"))) \
    .withColumn("days_between_booking_and_show", datediff(col("show_date"), col("booking_date")))

dates_df.select("booking_id", "booking_date", "show_date", "days_between_booking_and_show").show(5)
```

```python
duplicates_df = dates_df.groupBy("booking_id").count().filter(col("count") > 1)
duplicates_df.show()
```

```python
deduped_df = dates_df.dropDuplicates(["booking_id"])
print("Total rows after deduplication:", deduped_df.count())
deduped_df.show(5)
```
