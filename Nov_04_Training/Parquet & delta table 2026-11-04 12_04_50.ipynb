Part A — Retrieving and filtering data

```python
%sql
select * from fitness_memberships_demo;
```

```python
%sql
-- 2. Retrieve only member_name, city, and membership_plan
SELECT member_name, city, membership_plan 
FROM fitness_memberships_demo;
```

```python
%sql
-- 3. Display members from Hyderabad
SELECT * 
FROM fitness_memberships_demo 
WHERE city = 'Hyderabad';
```

```python
%sql
-- 4. Find members whose monthly fee is greater than ₹2,000
SELECT * 
FROM fitness_memberships_demo 
WHERE monthly_fee > 2000;
```

```python
%sql
-- 5. Display all distinct cities
SELECT DISTINCT city 
FROM fitness_memberships_demo;
```

```python
%sql
-- 6. Find members belonging to the Premium or Standard plan using IN
SELECT * 
FROM fitness_memberships_demo 
WHERE membership_plan IN ('Premium', 'Standard');
```

```python
%sql
-- 7. Retrieve members whose monthly fee is between ₹1,000 and ₹2,000
SELECT * 
FROM fitness_memberships_demo 
WHERE monthly_fee BETWEEN 1000 AND 2000;
```

```python
%sql
-- 8. Find members whose names start with S using LIKE
SELECT * 
FROM fitness_memberships_demo 
WHERE member_name LIKE 'S%';
```

```python
%sql
-- 9. Display Active members from Hyderabad using AND
SELECT * 
FROM fitness_memberships_demo 
WHERE status = 'Active' AND city = 'Hyderabad';
```

```python
%sql
-- 10. Find members from Pune or Chennai using OR
SELECT * 
FROM fitness_memberships_demo 
WHERE city = 'Pune' OR city = 'Chennai';
```

```python
%sql
-- 11. Display member names as Name and monthly fees as Fee using column aliases
SELECT member_name AS Name, monthly_fee AS Fee 
FROM fitness_memberships_demo;
```

```python
%sql
-- 12. Retrieve all Premium members using a table alias
SELECT f.* 
FROM fitness_memberships_demo AS f 
WHERE f.membership_plan = 'Premium';
```
