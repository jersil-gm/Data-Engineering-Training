**Aggregation operations**

```python
from pyspark.sql.functions import sum,avg,min,max,count,col
```

```python
#count
df.count()
```

```python
#sum
df.select(sum("unit_price").alias("Total_price")).show()
```

```python
#avg
df.select(avg("unit_price").alias("Average_price")).show()
```

```python
#max
df.select(max("unit_price").alias("Max_price")).show()
```

```python
# using multiple aggregate
df.select(sum("unit_price").alias("Total_price"),
          avg("unit_price").alias("Average_price"),
          max("unit_price").alias("Max_price")).show()
```

```python
# create gross_amount
df2=df.withColumn("gross_amount",col("quantity")*col("unit_price"))
df2.show()
```

```python
# Total gross sale
df2.select(sum("gross_amount").alias("Total_gross_sale")).show()
```

```python
# Create discount_amount
df2= df2.withColumn("discount_amount", col("gross_amount") * (col("discount_pct") / 100))
df2.select("gross_amount", "discount_pct", "discount_amount").show(5)
```

```python
# Find total discount given
total_discount = df2.select(sum("discount_amount").alias("total_discount_given"))
total_discount.show()
```

```python
# Create final_amount
df2 = df2.withColumn("final_amount", col("gross_amount") - col("discount_amount"))
df2.select("gross_amount", "discount_amount", "final_amount").show(5)
```

```python
# Find total final sales
total_final_sales = df2.select(sum("final_amount").alias("total_final_sales"))
total_final_sales.show()
```

```python
# Find total sales from Delivered orders only
delivered_sales = (
    df2.filter(col("status") == "Delivered")
      .select(sum("final_amount").alias("total_delivered_sales"))
)
delivered_sales.show()
```

```python
# Find the average value of Delivered orders
avg_delivered_order_val = (
    df2.filter(col("status") == "Delivered")
      .select(avg("final_amount").alias("avg_delivered_order_value"))
)
avg_delivered_order_val.show()
```

```python
# Count Electronics transactions
electronics_count = df.filter(col("category") == "Electronics").count()
print(electronics_count)
```

```python
# Count Cancelled transactions
cancelled_count = df.filter(col("status") == "Cancelled").count()
print("Cancelled transactions:",cancelled_count)
```

```python
# group by
df.groupBy("category").count().show()
```

```python
df.groupBy("status").count().show()
```

```python
# GroupBy and aggregate
df.groupBy("category").agg(sum("unit_price").alias("total_price"),
                           avg("unit_price").alias("avg_price"),
                           min("unit_price").alias("min_price"),
                           max("unit_price").alias("max_price"),
                           count("*").alias("Total_orders")).show()
```

**Joins**

```python
customers_data = [
    ("C00001", "Aman Retail", "Hyderabad", "Corporate"),
    ("C00002", "Blue Mart", "Mumbai", "Retail"),
    ("C00003", "Fresh Basket", "Bangalore", "Retail"),
    ("C00004", "Prime Stores", "Delhi", "SMB"),
    ("C00005", "Quick Shop", "Pune", "Corporate")
]

customers_columns = [
    "customer_id",
    "customer_name_master",
    "customer_city",
    "segment"
]

customers_df = spark.createDataFrame(
    customers_data,
    customers_columns
)

customers_df.show()
```

```python
# inner join
join_df=df.join(customers_df,on="customer_id",how="inner")
join_df.show()
```

```python
# Left Join
left_join_df=df.join(customers_df,on="customer_id",how="left")
left_join_df.show()
```

```python
# Right Join
right_join_df=df.join(customers_df,on="customer_id",how="right")
right_join_df.show()
```

```python
# Full Join
full_join_df=df.join(customers_df,on="customer_id",how="outer")
full_join_df.show()
```

**Window Function**

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import (
    col,
    row_number,
    rank,
    dense_rank,
    sum,
    avg,
    lag,
    lead
)
```

```python
window_spec = Window.partitionBy("category") \
    .orderBy(col("region").desc())

ranked_products = df.withColumn(
    "rank",
    row_number().over(window_spec)
)

ranked_products.show()
```

```python
ranked_products.filter(col("rank") <=3).show()
```

**Dates**

```python
from pyspark.sql.functions import col, to_date

df2 = df.withColumn(
    "order_date",
    to_date(col("order_date"), "yyyy-MM-dd")
)
```

```python
# To get year from date
from pyspark.sql.functions import year
df2.withColumn("year",year("order_date")).select("order_date","year").show()
```

```python
# To get month from date
from pyspark.sql.functions import month
df2.withColumn("month",month("order_date")).select("order_date","month").show()
```

```python
# To get day from the date
from pyspark.sql.functions import dayofmonth
df2.withColumn("Day",dayofmonth("order_date")).select("order_date","Day").show()
```
