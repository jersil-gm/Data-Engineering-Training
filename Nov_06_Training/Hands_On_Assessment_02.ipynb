## Section B — Filtering, joins and aggregations

```python
%sql
-- 7. Display the distinct registration channels.
SELECT DISTINCT channel 
FROM event_registrations_source;
```

```python
%sql
-- 8. Find attendees whose names begin with A or contain the letter a , using LIKE .
SELECT * 
FROM event_registrations_source 
WHERE attendee_name LIKE 'A%' OR attendee_name LIKE '%a%';
```

```python
%sql
-- 9. Calculate the total number of tickets registered for each event.
SELECT event_id, SUM(tickets) AS total_tickets 
FROM event_registrations_source 
GROUP BY event_id;
```

```python
%sql
-- 10. Calculate the total revenue of confirmed registrations for each event using tickets * unit_price . Exclude invalid records with zero tickets or zero price.
SELECT 
    event_id, 
    SUM(tickets * unit_price) AS total_revenue 
FROM event_registrations_source 
WHERE status = 'Confirmed' 
  AND tickets > 0 
  AND unit_price > 0 
GROUP BY event_id;
```

```python
%sql
-- 11. Join event_registrations_source and event_master to display attendee name, event name, city, ticket type, and registration status.
SELECT 
    r.attendee_name, 
    m.event_name, 
    m.city, 
    r.ticket_type, 
    r.status 
FROM event_registrations_source r 
JOIN event_master m 
  ON r.event_id = m.event_id;
```

```python
%sql
-- 12. Perform a left join and identify registrations whose event IDs do not exist in event_master . Also identify events with no registrations.

SELECT r.* 
FROM event_registrations_source r 
LEFT JOIN event_master m 
  ON r.event_id = m.event_id 
WHERE m.event_id IS NULL;


SELECT m.* 
FROM event_master m 
LEFT JOIN event_registrations_source r 
  ON m.event_id = r.event_id 
WHERE r.event_id IS NULL;
```

## Section C — PySpark DataFrames

```python
# 13. Load both Delta tables into separate PySpark DataFrames and display their schemas.

df_registrations = spark.read.table("event_registrations_source")
df_events = spark.read.table("event_master")

print("Schema of event_registrations_source:")
df_registrations.printSchema()

print("Schema of event_master:")
df_events.printSchema()
```

```python
# 14. Select only registration_id , attendee_name , tickets , and unit_price using PySpark.
df_selected = df_registrations.select("registration_id", "attendee_name", "tickets", "unit_price")
display(df_selected)
```

```python
# 15. Filter records where the status is Confirmed and the number of tickets is greater than 2.

df_filtered = df_registrations.filter((df_registrations.status == "Confirmed") & (df_registrations.tickets > 2))
display(df_filtered)
```

```python
# 16. Create a new column named total_amount by multiplying tickets and unit_price .
df_amount = df_registrations.withColumn("total_amount", df_registrations.tickets * df_registrations.unit_price)
display(df_amount)
```

```python
# 17. Create a column named booking_category with the values High for total amounts of ₹2,000 or more and Regular otherwise.
from pyspark.sql import functions as F

df_booking = df_amount.withColumn(
    "booking_category",
    F.when(F.col("total_amount") >= 2000, "High").otherwise("Regular")
)
display(df_booking)
```

```python
# 18. Perform a PySpark join between registrations and the event master. Group by city and calculate the total confirmed ticket revenue.
from pyspark.sql import functions as F

df_joined = df_registrations.filter(F.col("status") == "Confirmed") \
    .join(df_events, "event_id") \
    .groupBy("city") \
    .agg(F.sum(F.col("tickets") * F.col("unit_price")).alias("total_confirmed_revenue"))

display(df_joined)
```

## Section E — MERGE, UPSERT and SCD

```python
%sql
-- 25. Compare incoming_registrations with event_registrations_work . Identify existing and new registration IDs.
SELECT i.registration_id, 'Existing' AS record_type 
FROM incoming_registrations i 
INNER JOIN event_registrations_work w 
  ON i.registration_id = w.registration_id

UNION ALL

SELECT i.registration_id, 'New' AS record_type 
FROM incoming_registrations i 
LEFT JOIN event_registrations_work w 
  ON i.registration_id = w.registration_id 
WHERE w.registration_id IS NULL;
```

```python
%sql
-- 26. Perform a SQL MERGE that updates existing registrations and inserts new registrations.
MERGE INTO event_registrations_work AS target
USING incoming_registrations AS source
ON target.registration_id = source.registration_id
WHEN MATCHED THEN
  UPDATE SET
    target.event_id = source.event_id,
    target.attendee_name = source.attendee_name,
    target.registration_date = source.registration_date,
    target.ticket_type = source.ticket_type,
    target.tickets = source.tickets,
    target.unit_price = source.unit_price,
    target.status = source.status,
    target.channel = source.channel
WHEN NOT MATCHED THEN
  INSERT (registration_id, event_id, attendee_name, registration_date, ticket_type, tickets, unit_price, status, channel)
  VALUES (source.registration_id, source.event_id, source.attendee_name, source.registration_date, source.ticket_type, source.tickets, source.unit_price, source.status, source.channel);
```

```python
%sql
-- 38. Create a permanent SQL view named event_revenue_view containing event name, city, and total confirmed revenue.
CREATE OR REPLACE VIEW event_revenue_view AS
SELECT 
    m.event_name,
    m.city,
    SUM(r.tickets * r.unit_price) AS total_confirmed_revenue
FROM event_registrations_source r
JOIN event_master m
  ON r.event_id = m.event_id
WHERE r.status = 'Confirmed'
GROUP BY m.event_name, m.city;

SELECT * FROM event_revenue_view;
```

```python
%sql
-- 39. Create a temporary view containing only Pending registrations.
CREATE OR REPLACE TEMP VIEW pending_registrations_view AS
SELECT * 
FROM event_registrations_source 
WHERE status = 'Pending';

SELECT * FROM pending_registrations_view;
```
