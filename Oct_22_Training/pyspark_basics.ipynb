```python
from pyspark.sql.types import (
    StructType,
    StructField,
    IntegerType,
    StringType,
    DoubleType,
    BooleanType
)
```

**Exercise Set 4 — withColumn()**

```python
from pyspark.sql.functions import col

# 23. Add gross_amount = quantity * price
df = df.withColumn("gross_amount", col("quantity") * col("price"))
df.select("quantity", "price", "gross_amount").show(5)
```

```python
# 24. Create discount_amount = gross_amount * discount_pct / 100
df = df.withColumn("discount_amount", col("gross_amount") * col("discount_pct") / 100)
df.select("gross_amount", "discount_pct", "discount_amount").show(5)
```

```python
# 25. Create final_amount = gross_amount - discount_amount
df = df.withColumn("final_amount", col("gross_amount") - col("discount_amount"))
df.select("gross_amount", "discount_amount", "final_amount").show(5)
```

```python
# 26. Add a constant column: country = India
from pyspark.sql.functions import lit
df = df.withColumn("country", lit("India"))
df.select("country").show(5)
```

```python
# 27. Add high_value (final_amount >= 50000 -> Yes, else -> No)
from pyspark.sql.functions import col, when
df = df.withColumn(
    "high_value",
    when(col("final_amount") >= 50000, "Yes").otherwise("No")
)
df.select("final_amount", "high_value").show(5)
```

**Exercise Set 5 — Conditional Columns**

```python
# 28. Create order_size (>= 50000: High, >= 15000: Medium, otherwise: Low)
df = df.withColumn(
    "order_size",
    when(col("final_amount") >= 50000, "High")
    .when(col("final_amount") >= 15000, "Medium")
    .otherwise("Low")
)
df.select("final_amount", "order_size").show(5)
```

```python
# 29. Create quantity_type (quantity >= 4 -> Bulk, otherwise -> Regular)
df = df.withColumn(
    "quantity_type",
    when(col("quantity") >= 4, "Bulk").otherwise("Regular")
)
df.select("quantity", "quantity_type").show(5)
```

```python
# 30. Create delivery_status category (Delivered -> Completed, Cancelled/Returned -> Failed, Others -> In Progress)
df = df.withColumn(
    "delivery_status_category",
    when(col("status") == "Delivered", "Completed")
    .when(col("status").isin("Cancelled", "Returned"), "Failed")
    .otherwise("In Progress")
)
df.select("status", "delivery_status_category").show(5)
```
